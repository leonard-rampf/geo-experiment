```python
import numpy as np
from statsmodels.stats.power import TTestPower
import math

# 1. Define the a priori parameters
effect_size_dz = 0.15   # A conservatively small effect size
alpha = 0.05            # Standard 5% significance level
power = 0.80            # 80% statistical power

# 2. Calculate the required sample size for a Parametric Paired T-Test
# We use one-sample/paired T-test power solver
ttest_power_solver = TTestPower()
n_ttest = ttest_power_solver.solve_power(
    effect_size=effect_size_dz, 
    alpha=alpha, 
    power=power, 
    alternative='two-sided'
)

# 3. Apply the Pitman Asymptotic Relative Efficiency (A.R.E.) for Wilcoxon
# The A.R.E. of Wilcoxon to t-test under normality is 3/pi
are_adjustment = 3 / math.pi
n_wilcoxon = n_ttest / are_adjustment

# 4. Round up to the nearest whole integer (you cannot have a fraction of a query)
required_n = math.ceil(n_wilcoxon)

print(f"Target Effect Size (dz): {effect_size_dz}")
print(f"Alpha (Type I Error):    {alpha}")
print(f"Target Power (1 - Beta): {power}")
print("-" * 60)
print(f"Raw T-Test N:            {n_ttest:.2f}")
print(f"A.R.E. Adjustment Factor:{are_adjustment:.4f}")
print(f"Exact Wilcoxon N:        {n_wilcoxon:.2f}")
print("-" * 60)
print(f"REQUIRED SAMPLE SIZE:   {required_n} pairs")
```
