## Create selected_docs_seo_edits.json

Copies `selected_docs.json` into `selected_docs_seo_edits.json` for each domain.
Strips all existing manipulation keys — only keeps original document fields.

**Only run once per domain.** Skipped if file already exists.

````python
DOMAINS = ["retail", "videogames"]
KEYS_TO_KEEP = ["query_id", "query", "list_docs", "target_doc_idx", "doc"]

for domain in DOMAINS:
    src_path = os.path.join(data_dir, domain, "selected_docs.json")
    dst_path = os.path.join(data_dir, domain, "selected_docs_seo_edits.json")

    if os.path.exists(dst_path):
        print(f"[{domain}] selected_docs_seo_edits.json already exists — skipping.")
        continue

    if not os.path.exists(src_path):
        print(f"[{domain}] WARNING: selected_docs.json not found. Run Notebook 1 first.")
        continue

    with open(src_path, "r", encoding="utf-8") as f:
        docs = json.load(f)

    clean_docs = {}
    for idx, entry in docs.items():
        clean_docs[idx] = {k: entry[k] for k in KEYS_TO_KEEP if k in entry}

    with open(dst_path, "w", encoding="utf-8") as f:
        json.dump(clean_docs, f, indent=4, ensure_ascii=False)

    print(f"[{domain}] Created selected_docs_seo_edits.json ({len(clean_docs)} queries, no manipulation keys)")
````

## Parameters

````python
# LLM
LLM_NAME = "gpt-4o-mini-2024-07-18"

# <- CHANGE THIS to control which domains to process
# For both: DOMAINS = ["retail", "videogames"]
# For one:  DOMAINS = ["retail"]
DOMAINS = ["retail", "videogames"]

# Methods — all 11 edits
METHOD_NAMES = list(METHOD_CLASSES.keys())

# Calculate remaining calls
total_calls = 0
for domain in DOMAINS:
    path = os.path.join(data_dir, domain, "selected_docs_seo_edits.json")
    if os.path.exists(path):
        with open(path, "r") as f:
            docs = json.load(f)
        remaining = sum(
            len(METHOD_NAMES) - sum(1 for m in METHOD_NAMES if docs[idx].get(f"{m}(doc)") is not None)
            for idx in docs
        )
        total_calls += remaining

print(f"LLM:             {LLM_NAME}")
print(f"Domains:         {DOMAINS}")
print(f"Methods:         {METHOD_NAMES}")
print(f"Remaining calls: {total_calls}")
print(f"Estimated time:  {total_calls * 4 // 60}-{total_calls * 5 // 60} minutes")
print(f"Estimated cost:  ~${total_calls * 0.0003:.2f}")
````

## Manipulation

Runs all 11 isolated SEO edits for each domain.
Safe to interrupt and resume — already processed documents are skipped.

````python
llm = OpenAIHelper(LLM_NAME)
start_time = datetime.now()
print(f"Started at: {start_time.strftime('%H:%M:%S')}")
print()

for domain in DOMAINS:
    selected_docs_path = os.path.join(data_dir, domain, "selected_docs_seo_edits.json")

    if not os.path.exists(selected_docs_path):
        print(f"[{domain}] WARNING: selected_docs_seo_edits.json not found.")
        print(f"  Run the 'Create selected_docs_seo_edits.json' cell first.")
        continue

    with open(selected_docs_path, "r", encoding="utf-8") as f:
        selected_docs = json.load(f)

    query_indices = list(selected_docs.keys())
    total_docs = len(query_indices)

    print(f"{'='*60}")
    print(f"DOMAIN: {domain.upper()} ({total_docs} documents)")
    print(f"{'='*60}")

    for method_name in METHOD_NAMES:
        method = METHOD_CLASSES[method_name](llm)
        key = f"{method_name}(doc)"
        already_done = sum(1 for d in selected_docs.values() if d.get(key) is not None)

        if already_done == total_docs:
            print(f"  [{method_name}] All {total_docs} docs already done — skipping")
            continue

        print(f"  [{method_name}] Starting ({already_done}/{total_docs} already done)")

        for i, query_idx in enumerate(query_indices):
            if selected_docs[query_idx].get(key) is not None:
                continue

            doc = selected_docs[query_idx]["doc"]

            try:
                result = method.improve_text(doc)
                selected_docs[query_idx][key] = result
                with open(selected_docs_path, "w", encoding="utf-8") as f:
                    json.dump(selected_docs, f, indent=4, ensure_ascii=False)
                print(f"    [{i+1}/{total_docs}] {query_idx}: done")

            except Exception as e:
                print(f"    [{i+1}/{total_docs}] {query_idx}: ERROR — {e}")
                print(f"    Retrying in 10 seconds...")
                time.sleep(10)
                try:
                    result = method.improve_text(doc)
                    selected_docs[query_idx][key] = result
                    with open(selected_docs_path, "w", encoding="utf-8") as f:
                        json.dump(selected_docs, f, indent=4, ensure_ascii=False)
                    print(f"    [{i+1}/{total_docs}] {query_idx}: done (retry OK)")
                except Exception as e2:
                    print(f"    [{i+1}/{total_docs}] {query_idx}: FAILED — {e2}")
                    selected_docs[query_idx][key] = None

        done = sum(1 for d in selected_docs.values() if d.get(key) is not None)
        print(f"  [{method_name}] Complete: {done}/{total_docs}")

    print(f"  Domain '{domain}' done at {datetime.now().strftime('%H:%M:%S')}")
    print()

end_time = datetime.now()
elapsed = end_time - start_time
print(f"{'='*60}")
print(f"ALL DOMAINS COMPLETE")
print(f"Started:    {start_time.strftime('%H:%M:%S')}")
print(f"Finished:   {end_time.strftime('%H:%M:%S')}")
print(f"Total time: {str(elapsed).split('.')[0]}")
````
